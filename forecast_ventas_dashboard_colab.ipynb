# Forecast de Ventas + Dashboard

Proyecto end-to-end listo para **Google Colab**.

**Pregunta:** ¿cómo predecir ventas y visualizarlas?

**Método:** Regresión simple + gráficos

**Stack:** pandas, scikit-learn, matplotlib

Datos simulados.

## 0. Setup

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
print("OK")
```

## 1. Datos simulados

```python
np.random.seed(42)
meses = np.arange(1, 25)
ventas = 1000 + 50 * meses + np.random.normal(0, 100, 24)

df = pd.DataFrame({"mes": : meses, "ventas": : ventas})
print(df.head())
```

## 2. Modelo de forecast

```python
X = df[["mes"]]
y = df["ventas"]

model = LinearRegression()
model.fit(X, y)

# Predicción futuros meses
futuros = np.array([[25], [26], [27]])
preds = model.predict(futuros)
print("Predicciones:", preds)
```

## 3. Dashboard simple

```python
plt.figure(figsize=(10, 4))
plt.plot(df["mes"], df["ventas"], label="Histórico", marker="o")
plt.plot([25, 26, 27], preds, label="Forecast", marker="x", linestyle="--")
plt.title("Forecast de Ventas")
plt.xlabel("Mes")
plt.ylabel("Ventas")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

## 4. Conclusiones

- Un forecast simple permite planificar inventario y recursos.
- La visualización tipo dashboard facilita la comunicación.
- Se puede extender a Prophet, ARIMA o modelos más complejos.
